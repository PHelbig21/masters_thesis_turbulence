```python
N = 512
L = 2*np.pi
l_num_bins_per_decade = 7


num_bins = int(np.floor(np.log10(N/2) * l_num_bins_per_decade)) + 1 # number of decades * number of bins per decade
l_bin_centers = L/N * 10**(np.arange(num_bins) / l_num_bins_per_decade)
# construct inner bin boundaries
l_bins = np.zeros(num_bins+1)
l_bins[1:-1] = np.sqrt(l_bin_centers[:-1] * l_bin_centers[1:])
# first and last boundary
l_bins[0] = l_bin_centers[0]**2 / l_bins[1]
l_bins[-1] = l_bin_centers[-1]**2 / l_bins[-2]
# bin width vector
l_bin_width = np.diff(l_bins)



print(l_bins)
print(l_bin_centers)
```

```python
def toCartesian(phi, theta):
    x = np.cos(phi) * np.sin(theta)
    y = np.sin(phi) * np.sin(theta)
    z = np.cos(theta)
    return((x,y,z))
```

```python
phis = np.linspace(0, 2*np.pi, 6)
thetas = np.linspace(np.pi/4, 3*np.pi/4, 3)

phis, thetas = np.meshgrid(phis, thetas)
phis = phis.flatten()
thetas = thetas.flatten()

#print(phis)
#print(thetas)

xs, ys, zs = toCartesian(phis, thetas)
xyzs = np.array([xs, ys, zs])

print(xs)
print(xyzs)

#print(xs)
#print(ys)
#print(zs)

ephis = np.array([-np.sin(phis), np.cos(phis), np.zeros_like(phis)])
ethetas = np.array([np.cos(thetas)*np.cos(phis), np.cos(thetas)*np.sin(phis), -np.sin(thetas)])

#print(ephis)
#print(ethetas)
```

```python
uphis = np.random.rand(18)
uthetas = np.random.rand(18)

us = uphis*ephis + uthetas*ethetas

print(us)
```
